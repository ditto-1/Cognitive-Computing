Q1.  Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.).

1. Convert text to lowercase and remove punctua on using re.
2. Tokenize the text into words and sentences.

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Original paragraph
paragraph = (
    "The James Webb Space Telescope continues to revolutionize our understanding of the universe. "
    "By capturing infrared light from the earliest galaxies, it peers billions of years back in time. "
    "Astronomers are now able to analyze atmospheric compositions of distant exoplanets with unprecedented precision. "
    "These discoveries bridge theoretical astrophysics with tangible cosmic reality. "
    "Every breathtaking image reminds us of how much remains to be uncovered in the vast cosmos."
)

# 1. Convert text to lowercase and remove punctuation using re
import re
text_lower = paragraph.lower()
text_clean = re.sub(r'[^\w\s]', '', text_lower)

# 2. Tokenize into sentences and words
sentences = sent_tokenize(paragraph)
words_nltk = word_tokenize(text_clean)

print("Tokenization successful!")
print("Sentences:", len(sentences))
print("Words:", len(words_nltk))
```

Q2. Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall()
2. Remove stop words using NLTK’s stopword list
3. Perform stemming with PorterStemmer
4. Perform lemma za on with WordNetLemma zer
5. Compare the stemmed and lemma zed outputs and explain when you’d prefer one over the other.

```python
stop_words = set(stopwords.words('english'))

# 1. Extract all words with only alphabets using re.findall()
alpha_words = re.findall(r'\b[a-zA-Z]+\b', paragraph.lower())

# 2. Remove stop words using NLTK's stopword list
clean_alpha_words = [w for w in alpha_words if w not in stop_words]

# 3. Perform stemming with PorterStemmer
porter = PorterStemmer()
stemmed_words = [porter.stem(w) for w in clean_alpha_words]

# 4. Perform lemmatization with WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(w) for w in clean_alpha_words]

# 5. Compare outputs
print("--- Q2: Stemming vs Lemmatization Comparison ---")
print(f"{'Original':<15} | {'Stemmed':<12} | {'Lemmatized'}")
print("-" * 45)
for original, stemmed, lemmatized in list(zip(clean_alpha_words, stemmed_words, lemmatized_words))[:8]:
    print(f"{original:<15} | {stemmed:<12} | {lemmatized}")
```

Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representa on.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 3 short texts (Product reviews)
texts = [
    "The camera quality on this smartphone is absolutely stunning, especially in low light conditions.",
    "Battery life is terrible; the phone dies before the end of the workday even with minimal use.",
    "An incredible smartphone with a brilliant display, blazing fast performance, and solid battery life."
]

# 1. CountVectorizer (Bag of Words representation)
count_vec = CountVectorizer(stop_words='english')
bow_matrix = count_vec.fit_transform(texts)
bow_df = pd.DataFrame(bow_matrix.toarray(), columns=count_vec.get_feature_names_out())

# 2. TfidfVectorizer (TF-IDF scores)
tfidf_vec = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vec.fit_transform(texts)
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vec.get_feature_names_out())

# 3. Print and interpret top 3 keywords from each text using TF-IDF
feature_names = tfidf_vec.get_feature_names_out()
for idx, row in tfidf_df.iterrows():
    top_3_indices = row.to_numpy().argsort()[-3:][::-1]
    top_keywords = [(feature_names[i], round(row[i], 4)) for i in top_3_indices]
    print(f"Text {idx + 1}: {top_keywords}")
```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:
*   Jaccard Similarity using sets
*   Cosine Similarity using TfidfVectorizer + cosine_similarity()
*   Analyze which similarity metric gives be er insights in your case.





```python
# 2 short texts describing two different technologies
tech_ai = (
    "Artificial intelligence empowers machines to mimic human cognition and learning. "
    "Neural networks process massive datasets to recognize complex patterns. "
    "Deep learning models drive breakthroughs in computer vision and natural language processing. "
    "Automation transforms industries by optimizing workflows and decision-making."
)

tech_blockchain = (
    "Blockchain technology establishes a decentralized and immutable digital ledger. "
    "Cryptographic hashing secures transactions across distributed peer-to-peer networks. "
    "Smart contracts execute automated agreements without relying on trusted intermediaries. "
    "Distributed consensus ensures high transparency and unalterable data integrity."
)

# 1. Preprocess and tokenize
def get_alpha_set(text):
    return set(re.findall(r'\b[a-zA-Z]+\b', text.lower())) - stop_words

set_ai = get_alpha_set(tech_ai)
set_blockchain = get_alpha_set(tech_blockchain)

# 2a. Jaccard Similarity using sets
intersection = len(set_ai.intersection(set_blockchain))
union = len(set_ai.union(set_blockchain))
jaccard_sim = intersection / union if union > 0 else 0

# 2b. Cosine Similarity using TfidfVectorizer + cosine_similarity()
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform([tech_ai, tech_blockchain])
cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])[0][0]

print(f"Jaccard Similarity: {jaccard_sim:.4f}")
print(f"Cosine Similarity: {cosine_sim:.4f}")
```
