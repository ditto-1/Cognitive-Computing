iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension

```python
import math
def isPrime(n):
  if n < 1:
    return False
  elif n == 2:
    return True
  elif n%2 == 0:
    return False

  for x in range(3, int(math.sqrt(n))+1, 2):
    if n%x == 0:
      return False
  return True

prime = [x for x in numbers if isPrime(x)]
print(prime)
print("Total prime: ", len(prime))
```
