b) Find the most frequent value and their indice(s) in the following arrays: 

i. x = np.array([1,2,3,4,5,1,2,1,1,1]) 

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
value, counts = np.unique(x, return_counts=True)
most_freq_value = value[np.argmax(counts)]
indices = np.argwhere(x == most_freq_value)
print("Most frequent value in x: ", most_freq_value)
print("At indices: ", indices.flatten())
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
value, counts = np.unique(y, return_counts=True)
max_Count = np.max(counts)
most_freq_value = value[counts == max_Count]
for i in most_freq_value:
    print("Most frequent value: ", i)
    indices = np.argwhere(y == i)
    print("Indices: ", indices.flatten())
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute? 

```python
deeksha = np.linspace(10, 100, 25)
print(deeksha.ndim)
print(deeksha.shape)
print(deeksha.size)
print(deeksha.dtype)
print(deeksha.nbytes)

transpose = deeksha.reshape(-1, 1)
print("Transpose: ", transpose)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>

```python
ucs420_deeksha = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print(f"Mean: {np.mean(ucs420_deeksha)}")
print(f"Median: {np.median(ucs420_deeksha)}")
print(f"Max: {np.max(ucs420_deeksha)}")
print(f"Min: {np.min(ucs420_deeksha)}")
print(f"Unique Elements: {np.unique(ucs420_deeksha)}")

reshaped_ucs402_deeksha = ucs420_deeksha.reshape(4, 3)
print("Reshaped: ")
print(reshaped_ucs402_deeksha)
resized_ucs420_deeksha = np.resize(ucs420_deeksha, (2, 3))
print("Resized:")
print(resized_ucs420_deeksha)
```
